```python
# ── Milestone 6: Comparison & Analysis ───────────────────────────────────────
# Final head-to-head comparison of GARCH vs LSTM.
# ROC curves, confusion matrices, regime prediction timeline,
# and sensitivity check with alternative VIX threshold.

import pandas as pd
import numpy as np
import torch
import torch.nn as nn
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
from torch.utils.data import Dataset, DataLoader
from sklearn.metrics import (accuracy_score, precision_score,
                             recall_score, f1_score, roc_auc_score,
                             confusion_matrix, ConfusionMatrixDisplay,
                             RocCurveDisplay)
import json, os
```

```python
# ── Reproducibility ────────────────────────────────────────────────────────────
SEED = 42
torch.manual_seed(SEED)
np.random.seed(SEED)

os.makedirs("plots", exist_ok=True)
```

```python
# ── Load saved results ─────────────────────────────────────────────────────────
garch_results = pd.read_csv("data/processed/garch_results.csv",
                            index_col="date", parse_dates=True)
lstm_results  = pd.read_csv("data/processed/lstm_results.csv",
                            index_col="date", parse_dates=True)
cv_summary    = pd.read_csv("data/processed/cv_summary.csv")
cv_results    = pd.read_csv("data/processed/cv_results.csv")
df_master     = pd.read_csv("data/processed/master.csv",
                            index_col="date", parse_dates=True)

print("GARCH results shape :", garch_results.shape)
print("LSTM results shape  :", lstm_results.shape)
```

```python
# ── Align on common test dates ─────────────────────────────────────────────────
common_dates  = garch_results.index.intersection(lstm_results.index)
garch_aligned = garch_results.loc[common_dates]
lstm_aligned  = lstm_results.loc[common_dates]

actual        = lstm_aligned["actual_regime"].values
garch_preds   = garch_aligned["garch_pred"].values
garch_probs   = garch_aligned["garch_vol"].values
lstm_preds    = lstm_aligned["lstm_pred"].values
lstm_probs    = lstm_aligned["lstm_prob"].values

print(f"\nCommon test period : {common_dates[0].date()} "
      f"to {common_dates[-1].date()}")
print(f"Test days          : {len(common_dates)}")
```

```python
# ── 1. Final head-to-head metrics table ───────────────────────────────────────
def compute_metrics(labels, preds, probs, name):
    return {
        "Model":     name,
        "Accuracy":  round(accuracy_score(labels, preds), 4),
        "Precision": round(precision_score(labels, preds, zero_division=0), 4),
        "Recall":    round(recall_score(labels, preds, zero_division=0), 4),
        "F1 Score":  round(f1_score(labels, preds, zero_division=0), 4),
        "ROC-AUC":   round(roc_auc_score(labels, probs), 4),
    }

garch_m = compute_metrics(actual, garch_preds, garch_probs, "GARCH(1,1)")
lstm_m  = compute_metrics(actual, lstm_preds,  lstm_probs,  "LSTM")

final_metrics = pd.DataFrame([garch_m, lstm_m])
print(f"\n{'='*60}")
print("Final Head-to-Head Metrics (held-out test set)")
print('='*60)
print(final_metrics.to_string(index=False))

print(f"\n{'='*60}")
print("Walk-Forward CV Summary (mean ± std across folds)")
print('='*60)
print(cv_summary.to_string(index=False))
```

```python
# ── 2. ROC curves — both models on same axes ───────────────────────────────────
fig, ax = plt.subplots(figsize=(7, 6))
RocCurveDisplay.from_predictions(actual, garch_probs,
                                  name="GARCH(1,1)", ax=ax,
                                  color="steelblue")
RocCurveDisplay.from_predictions(actual, lstm_probs,
                                  name="LSTM", ax=ax,
                                  color="orange")
ax.plot([0, 1], [0, 1], linestyle="--", color="gray",
        label="Random classifier")
ax.set_title("ROC Curves: GARCH vs LSTM (Test Set)")
ax.legend(loc="lower right")
plt.tight_layout()
plt.savefig("plots/final_roc_comparison.png", dpi=150, bbox_inches="tight")
plt.show()
```

```python
# ── 3. Confusion matrices side by side ────────────────────────────────────────
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
for ax, preds, title in zip(axes,
                             [garch_preds, lstm_preds],
                             ["GARCH(1,1)", "LSTM"]):
    cm   = confusion_matrix(actual, preds)
    disp = ConfusionMatrixDisplay(confusion_matrix=cm,
                                  display_labels=["Calm", "Turbulent"])
    disp.plot(ax=ax, colorbar=False, cmap="Blues")
    ax.set_title(f"{title} Confusion Matrix")

plt.suptitle("Confusion Matrices: GARCH vs LSTM", fontsize=13, y=1.02)
plt.tight_layout()
plt.savefig("plots/final_confusion_matrices.png", dpi=150, bbox_inches="tight")
plt.show()
```

```python
# ── 4. Regime prediction timeline ─────────────────────────────────────────────
# Show where each model succeeds and fails — especially around crisis periods
fig, axes = plt.subplots(3, 1, figsize=(16, 10), sharex=True)

# Panel 1: Actual regime with VIX overlay
raw_vix = df_master["vix_diff"].cumsum()   # reconstruct approximate VIX level
axes[0].fill_between(common_dates, 0, 1,
                     where=(actual == 1),
                     alpha=0.3, color="red",
                     transform=axes[0].get_xaxis_transform(),
                     label="Actual turbulent")
axes[0].set_ylabel("Regime")
axes[0].set_title("Actual Regime")
axes[0].legend(fontsize=9)

# Panel 2: GARCH predictions vs actual
axes[1].step(common_dates, actual,      color="black",
             linewidth=0.8, label="Actual",  where="post")
axes[1].step(common_dates, garch_preds, color="steelblue",
             linewidth=0.8, alpha=0.8, label="GARCH Pred", where="post")
axes[1].set_yticks([0, 1])
axes[1].set_yticklabels(["Calm", "Turbulent"])
axes[1].set_ylabel("Regime")
axes[1].set_title("GARCH(1,1) Predictions vs Actual")
axes[1].legend(fontsize=9)

# Panel 3: LSTM predictions vs actual
axes[2].step(common_dates, actual,     color="black",
             linewidth=0.8, label="Actual", where="post")
axes[2].step(common_dates, lstm_preds, color="orange",
             linewidth=0.8, alpha=0.8, label="LSTM Pred", where="post")
axes[2].set_yticks([0, 1])
axes[2].set_yticklabels(["Calm", "Turbulent"])
axes[2].set_ylabel("Regime")
axes[2].set_title("LSTM Predictions vs Actual")
axes[2].legend(fontsize=9)
axes[2].xaxis.set_major_locator(mdates.YearLocator(1))
axes[2].xaxis.set_major_formatter(mdates.DateFormatter("%Y"))
axes[2].tick_params(axis="x", rotation=45)

plt.suptitle("Regime Prediction Timeline: GARCH vs LSTM",
             fontsize=14, y=1.01)
plt.tight_layout()
plt.savefig("plots/final_timeline.png", dpi=150, bbox_inches="tight")
plt.show()
```

```python
# ── 5. Sensitivity check — alternative VIX threshold ─────────────────────────
# Does the conclusion change if we use VIX=25 instead of VIX=20?
print(f"\n{'='*60}")
print("Sensitivity Check: VIX Threshold = 25")
print('='*60)

# Reload raw VIX from master to reconstruct labels at threshold=25
# We need the original (non-differenced) VIX — load from raw
raw_vix_series = pd.read_csv("data/raw/vix.csv",
                              index_col="date", parse_dates=True)
raw_vix_series = raw_vix_series.reindex(common_dates).ffill()

# New labels at threshold=25
actual_25 = (raw_vix_series["vix"].shift(-1) > 25).astype(int).dropna()
actual_25 = actual_25.reindex(common_dates).fillna(0).astype(int).values

sensitivity_rows = []
for threshold in [20, 25]:
    labels = actual if threshold == 20 else actual_25
    for name, preds, probs in [
        ("GARCH", garch_preds, garch_probs),
        ("LSTM",  lstm_preds,  lstm_probs)
    ]:
        vol_thresh = threshold / 100   # convert to decimal vol
        if name == "GARCH":
            preds_t = (garch_probs > vol_thresh).astype(int)
        else:
            preds_t = preds   # LSTM preds don't change with threshold

        sensitivity_rows.append({
            "VIX Threshold": threshold,
            "Model":         name,
            "F1 Score":      round(f1_score(labels, preds_t,
                                            zero_division=0), 4),
            "ROC-AUC":       round(roc_auc_score(labels, probs), 4),
            "Recall":        round(recall_score(labels, preds_t,
                                                zero_division=0), 4),
        })

sensitivity_df = pd.DataFrame(sensitivity_rows)
print(sensitivity_df.to_string(index=False))
```

```python
# ── 6. Crisis period breakdown ────────────────────────────────────────────────
# How well does each model do specifically during known crisis windows?
print(f"\n{'='*60}")
print("Performance During Known Crisis Periods")
print('='*60)

crisis_periods = {
    "GFC 2008":        ("2008-01-01", "2009-06-30"),
    "COVID 2020":      ("2020-02-01", "2020-12-31"),
    "Rate Shock 2022": ("2022-01-01", "2022-12-31"),
}

crisis_rows = []
for crisis, (start, end) in crisis_periods.items():
    mask = (common_dates >= start) & (common_dates <= end)
    if mask.sum() == 0:
        continue
    for name, preds, probs in [
        ("GARCH", garch_preds, garch_probs),
        ("LSTM",  lstm_preds,  lstm_probs)
    ]:
        sub_labels = actual[mask]
        sub_preds  = preds[mask]
        sub_probs  = probs[mask]
        if len(np.unique(sub_labels)) < 2:
            auc = float("nan")
        else:
            auc = round(roc_auc_score(sub_labels, sub_probs), 4)
        crisis_rows.append({
            "Crisis Period": crisis,
            "Model":         name,
            "F1":            round(f1_score(sub_labels, sub_preds,
                                            zero_division=0), 4),
            "Recall":        round(recall_score(sub_labels, sub_preds,
                                                zero_division=0), 4),
            "ROC-AUC":       auc,
            "N Days":        int(mask.sum()),
        })

crisis_df = pd.DataFrame(crisis_rows)
print(crisis_df.to_string(index=False))
```

```python
# ── 7. Final interpretation ───────────────────────────────────────────────────
print(f"\n{'='*60}")
print("Interpretation")
print('='*60)

lstm_f1   = lstm_m["F1 Score"]
garch_f1  = garch_m["F1 Score"]
lstm_auc  = lstm_m["ROC-AUC"]
garch_auc = garch_m["ROC-AUC"]
delta_f1  = lstm_f1 - garch_f1
delta_auc = lstm_auc - garch_auc

print(f"  LSTM F1   : {lstm_f1:.4f}")
print(f"  GARCH F1  : {garch_f1:.4f}")
print(f"  Delta F1  : {delta_f1:+.4f}")
print(f"  LSTM AUC  : {lstm_auc:.4f}")
print(f"  GARCH AUC : {garch_auc:.4f}")
print(f"  Delta AUC : {delta_auc:+.4f}")

if delta_f1 > 0.02:
    print("\n  ✅ POSITIVE RESULT: LSTM meaningfully outperforms GARCH.")
    print("     Nonlinear sequence structure adds predictive value beyond")
    print("     GARCH's autoregressive variance model.")
elif delta_f1 < -0.02:
    print("\n  📊 NULL/NEGATIVE RESULT: GARCH outperforms LSTM.")
    print("     Classical volatility modeling is sufficient for regime")
    print("     classification; deep learning does not add value here.")
else:
    print("\n  ⚖️  INCONCLUSIVE: Models perform comparably.")
    print("     Marginal difference may not be statistically meaningful.")
    print("     GARCH remains preferable given its interpretability.")
```

```python
# ── Save final outputs ─────────────────────────────────────────────────────────
final_metrics.to_csv("data/processed/final_metrics.csv", index=False)
sensitivity_df.to_csv("data/processed/sensitivity_results.csv", index=False)
crisis_df.to_csv("data/processed/crisis_breakdown.csv", index=False)
print(f"\nAll results saved to data/processed/")
print("Project complete.")
```
